## 7. Basic Dataset Facts

Explore key categorical dimensions: unique products, unique markets, available units, and price types.

## 8. Initial Observations

Summary of facts discovered directly from the dataset. No cleaning, transformations, or commodity/market conclusions are made at this stage.

```python
rows, cols = df.shape
n_products  = df["product"].nunique()
n_markets   = df["market"].nunique()
date_min    = df["period_date"].min()
date_max    = df["period_date"].max()
units       = df["unit"].unique().tolist()
price_types = df["price_type"].unique().tolist()
total_miss  = df.isna().sum().sum()

print("=" * 52)
print("  INITIAL DATASET OBSERVATIONS")
print("=" * 52)
print(f"  Rows              : {rows:,}")
print(f"  Columns           : {cols}")
print(f"  Unique products   : {n_products}")
print(f"  Unique markets    : {n_markets}")
print(f"  Date range        : {date_min}  to  {date_max}")
print(f"  Units available   : {units}")
print(f"  Price types       : {price_types}")
print(f"  Total missing vals: {total_miss:,}")
print("=" * 52)
```
